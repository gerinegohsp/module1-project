```python
import pandas as pd

# Load a 50,000-row sample first (full file has ~1M rows)
df = pd.read_csv('SGJobData.csv', nrows=50000)

print(df.shape)   # (rows, columns)
df.info()         # column names, types, missing values
df.head()         # preview first 5 rows
```

```python
# Summary statistics for the salary columns
df[['salary_minimum', 'salary_maximum', 'average_salary']].describe().round(0)
```

```python
# Distribution of seniority levels and employment types
print(df['positionLevels'].value_counts())
print()
print(df['employmentTypes'].value_counts())
```

```python
# Convert posting date from text to datetime, then check the range
df['metadata_originalPostingDate'] = pd.to_datetime(df['metadata_originalPostingDate'])
print('Earliest posting:', df['metadata_originalPostingDate'].min())
print('Latest posting:', df['metadata_originalPostingDate'].max())
```

# Module 1 Project — Singapore Jobs Analytics

## 1. Business Case

- **Business scenario:** A talent acquisition (TA) team at a mid-sized
  Singapore company planning its hiring strategy for the year ahead.
- **Objective:** Help the TA team decide which roles to prioritise, what
  salary ranges to offer, and when to post jobs — by revealing market
  demand, salary benchmarks, and competition levels across roles and
  industries.
- **Target users & value:** TA specialists and hiring managers. Instead of
  guessing salary bands or posting jobs blindly, they can benchmark against
  1M+ real MyCareersFuture postings — e.g. spotting roles where postings
  attract very few applications (hard-to-fill) vs. roles flooded with
  applicants (can hire selectively).

*Success criteria: a TA user can answer "what salary should we offer for
role X?" in under one minute using the dashboard.*
